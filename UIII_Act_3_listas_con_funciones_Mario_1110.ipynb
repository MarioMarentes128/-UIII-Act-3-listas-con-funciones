### Lista Youtuber

```python
# --- PROGRAMA DE GESTIÓN DE YOUTUBERS ---

youtubers = []

def agregar_youtuber():
    print("\n--- Agregar Nuevo Youtuber ---")
    nombre = input("Nombre del canal: ")
    tematica = input("Temática (Ej: Gaming, Educación): ")
    # Creamos un diccionario para el youtuber
    nuevo_youtuber = {"nombre": nombre, "tematica": tematica}
    youtubers.append(nuevo_youtuber)
    print(f"✅ ¡{nombre} ha sido agregado con éxito!")

def listar_youtubers():
    print("\n--- Lista de Youtubers ---")
    if not youtubers:
        print("La lista está vacía.")
    else:
        for i, y in enumerate(youtubers, start=1):
         print(f"{i}. Canal: {y['nombre']} | Temática: {y['tematica']}")

def buscar_youtuber():
    print("\n--- Buscar Youtuber ---")
    nombre_buscar = input("Ingresa el nombre del canal a buscar: ")
    encontrado = False
    for y in youtubers:
        if y['nombre'].lower() == nombre_buscar.lower():
            print(f"🔍 Encontrado: {y['nombre']} - Temática: {y['tematica']}")
            encontrado = True
            break
    if not encontrado:
        print("❌ No se encontró ningún youtuber con ese nombre.")

def modificar_youtuber():
    print("\n--- Modificar Youtuber ---")
    nombre_mod = input("Nombre del canal que deseas modificar: ")
    for y in youtubers:
        if y['nombre'].lower() == nombre_mod.lower():
            print(f"Datos actuales: {y['nombre']} ({y['tematica']})")
            y['nombre'] = input("Nuevo nombre: ")
            y['tematica'] = input("Nueva temática: ")
            print("📝 ¡Datos actualizados correctamente!")
            return
    print("❌ El Youtuber no existe.")

def borrar_youtuber():
    print("\n--- Borrar Youtuber ---")
    nombre_del = input("Nombre del canal a eliminar: ")
    for y in youtubers:
        if y['nombre'].lower() == nombre_del.lower():
            youtubers.remove(y)
            print(f"🗑️ El canal '{nombre_del}' ha sido eliminado.")
            return
    print("❌ No se pudo borrar: Canal no encontrado.")

# --- CICLO PRINCIPAL (MENÚ INTERACTIVO) ---

while True:
    print("\n============================")
    print("   MENÚ GESTIÓN YOUTUBERS   ")
    print("============================")
    print("1. Agregar Youtuber")
    print("2. Listar todos")
    print("3. Buscar por nombre")
    print("4. Modificar datos")
    print("5. Borrar Youtuber")
    print("6. Salir")

    opcion = input("\nSelecciona una opción (1-6): ")

    if opcion == "1":
        agregar_youtuber()
    elif opcion == "2":
        listar_youtubers()
    elif opcion == "3":
        buscar_youtuber()
    elif opcion == "4":
        modificar_youtuber()
    elif opcion == "5":
        borrar_youtuber()
    elif opcion == "6":
        print("¡Saliendo del programa... Adiós! 👋")
        break
    else:
        print("⚠️ Opción no válida, intenta de nuevo.")
```
